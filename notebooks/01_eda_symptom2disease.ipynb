```python
# EDA for Symptom2Disease dataset

from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure plots render nicely
sns.set(style="whitegrid")

# Build a robust absolute path regardless of notebook working directory
PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATA_PATH = PROJECT_ROOT / "src" / "data" / "raw" / "Symptom2Disease.csv"

# 1. Load data
print("Loading data from:", DATA_PATH)
if not DATA_PATH.exists():
    raise FileNotFoundError(f"Could not find dataset at: {DATA_PATH}")

df = pd.read_csv(DATA_PATH)
print("Shape:", df.shape)
df.head()
```

```python
# 2. Basic info and missing values

print("\nData types and non-null counts:")
df.info()

print("\nMissing values per column (top 20):")
print(df.isna().sum().sort_values(ascending=False).head(20))
```

```python
# 3. Target column and class balance

# TODO: set this to your real target column name
TARGET_COL = "Disease"  # change if needed

if TARGET_COL in df.columns:
    print("Target value counts:")
    print(df[TARGET_COL].value_counts())
    plt.figure(figsize=(10, 4))
    sns.countplot(y=TARGET_COL, data=df, order=df[TARGET_COL].value_counts().index)
    plt.title("Class distribution")
    plt.tight_layout()
else:
    print(f"TARGET_COL '{TARGET_COL}' not found in columns. Update it after inspecting df.head().")
```

```python
# 4. Basic correlations for numeric columns (if any)

numeric_cols = df.select_dtypes(include=["number"]).columns
print("Numeric columns:", list(numeric_cols))

if len(numeric_cols) > 1:
    plt.figure(figsize=(10, 8))
    corr = df[numeric_cols].corr()
    sns.heatmap(corr, annot=False, cmap="coolwarm", square=True)
    plt.title("Correlation heatmap (numeric features)")
    plt.tight_layout()
else:
    print("Not enough numeric columns for a correlation heatmap.")
```
